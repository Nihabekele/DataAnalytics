# Exercise 1.C Performing operations on arrays

```python
 import numpy as np
```

## Sample Arrays

```python
# Re-creating variables r_1 through r_5 from Exercise 1.B
r_1 = np.random.randint(1, 21, size=2)
r_2 = np.random.randint(1, 21, size=(1, 2))
r_3 = np.random.random(size=(2, 1))
r_4 = np.random.randint(20, 101, size=(4, 5))
r_5 = np.random.random(size=(4, 5))

print("r_1:\n", r_1)
print("\nr_2:\n", r_2)
print("\nr_3:\n", r_3)
print("\nr_4:\n", r_4)
print("\nr_5:\n", r_5)
```

## Permutations

```python
# 1. Print current value of r_4
print("Original r_4:\n", r_4)
```

```python
# 2. Shuffle r_4 and print again
np.random.shuffle(r_4)
print("\nShuffled r_4:\n", r_4)
```

```python
# When np.random.shuffle() is applied to a two-dimensional array like r_4, it only shuffles the array along its first axis (axis=0). This means the 
#rows themselves changed their order but the individual numbers inside each row stayed exactly the same and moved together.
```

```python
# 4. Create new_shuffle using permutation
new_shuffle = np.random.permutation(r_4)

print("new_shuffle (Permutation Result):\n", new_shuffle)
print("\nr_4 (Original Array after permutation):\n", r_4)
```

```python
## When to use Shuffle vs. Permutation

# Use np.random.shuffle() when I want to save computer memory and It does not matter to keeping the original order of my dataset. It 
# overwrites the data directly.
# Use np.random.permutation() when I need to keep my original dataset safe and unchanged because it outputs a brand new copy.
```

## Mathematical Operations

```python
print("r_1:\n", r_1, "Shape:", r_1.shape)
print("r_2:\n", r_2, "Shape:", r_2.shape)

# Adding r_1 and r_2
result_1_2 = np.add(r_1, r_2)
print("\nResult (r_1 + r_2):\n", result_1_2)
print("Result Shape:", result_1_2.shape)
```

```python
print("r_1:\n", r_1, "Shape:", r_1.shape)
print("r_3:\n", r_3, "Shape:", r_3.shape)

# Adding r_1 and r_3
result_1_3 = np.add(r_1, r_3)
print("\nResult (r_1 + r_3):\n", result_1_3)
print("Result Shape:", result_1_3.shape)
```

```python
print("r_2:\n", r_2, "Shape:", r_2.shape)
print("r_4:\n", r_4, "Shape:", r_4.shape)

# Adding r_2 and r_4
result_2_4 = np.add(r_2, r_4)
print("\nResult (r_2 + r_4):\n", result_2_4)
print("Result Shape:", result_2_4.shape)
```

```python
# 1. Expression to illustrate adding r_1 and r_4
np.add(r_1, r_4)
# error message
# ValueError: operands could not be broadcast together with shapes (2,) (4,5)

# 2. What happens when adding r_2 to r_4?
np.add(r_2, r_4)
# Error message - ValueError operands could not be broadcast together with shapes (1,2) (4,5)

# 3. What happens when adding r_3 to r_4?
np.add(r_3, r_4)
# Error message: ValueError: operands could not be broadcast together with shapes (2,1) (4,5)

# My Findings Comment
# All three combinations fail and give a ValueError. 
# For r_1 and r_2, their column size 2 does not match r_4's column size 5. 
# For r_3, even though its 1 column can stretch to match 5 columns its row size 2 does not match r_4's row size 4, so they cannot broadcast together.
```

```python
# Create the small array and add it to r_4
small_array_1 = np.array([1])
result_small_1 = np.add(r_4, small_array_1)

# Print the result
print("Result of r_4 + np.array([1]):")
print(result_small_1)
```

```python
# Create the 2D small array and add it to r_4
small_array_2 = np.array([[1]])
result_small_2 = np.add(r_4, small_array_2)

# Print the result
print("Result of r_4 + np.array([[1]]):")
print(result_small_2)
```

```python
# Create the 1D array with 5 elements and add it to r_4
row_array = np.array([1, 1, 1, 1, 1])
result_row = np.add(r_4, row_array)

# Print the result
print("Result of r_4 + np.array([1,1,1,1,1]):")
print(result_row)
```

```python
# Try to add the array to r_4 
np.add(r_4, np.array([[1], [1]]))

# Comment with the error message
# ValueError - operands could not be broadcast together with shapes (4,5) (2,1)
```

## How NumPy Broadcasting Works

How it works Broadcasting allows NumPy to do math on arrays of different sizes. It automatically stretches the smaller array so it matches the shape of the larger one.

What gets added to - what NumPy adds numbers element by element. If you add a single number [1] to a large array, that 1 gets added to every single individual number inside the large array.

What determines the final shape - The final shape takes the maximum size of the two arrays along each dimension (rows and columns) after the smaller dimensions of 1 are stretched to match the larger ones.


# Rounding

```python
# Print starting values of r_3
print("Original r_3 values")
print(r_3)
```

```python
# Round using trunc 
trunc_result = np.trunc(r_3)
print("Truncated values")
print(trunc_result)
```

```python
# Round using fix 
fix_result = np.fix(r_3)
print("Fixed values")
print(fix_result)
```

```python
# Round using floor 
floor_result = np.floor(r_3)
print("Floored values")
print(floor_result)
```

```python
# Round using ceil
ceil_result = np.ceil(r_3)
print("Ceiled values")
print(ceil_result)
```

```python
# Round to 2 decimal places
round_result = np.round(r_3, decimals=2)
print("Rounded to 2 decimal places")
print(round_result)
```

 ## Cumulative Operations

```python
# Print starting values of r_4
print("Original r_4 values")
print(r_4)
```

```python
# Sum of all vallue
total_sum = np.sum(r_4)
print("Sum of all values")
print(total_sum)
```

```python
# Sum down the columns 
sum_axis_0 = np.sum(r_4, axis=0)
print("Sum along axis 0")
print(sum_axis_0)
```

```python
# Sum across the rows 
sum_axis_1 = np.sum(r_4, axis=1)
print("Sum along axis 1")
print(sum_axis_1)
```

```python
# rolling total of all numbers
cumsum_flat = np.cumsum(r_4)
print("Cumulative sum")
print(cumsum_flat)
```

```python
# Rolling total adding down each column
cumsum_axis_0 = np.cumsum(r_4, axis=0)
print("Cumulative sum along axis 0")
print(cumsum_axis_0)
```

## Grocery Problem

```python
# Order of categories [Vegetables, Fruits, Snacks, Beverages]
bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])

print("bag1 contents:", bag1)
print("bag2 contents:", bag2)
```

```python
# Calculate the total items for each bag using sum
total_bag1 = np.sum(bag1)
total_bag2 = np.sum(bag2)

# Print the result 
print(f"The first bag contains {total_bag1} items, and the second bag contains {total_bag2} items.")
```

```python
# Add the bags together to get totals by category
category_totals = np.add(bag1, bag2)

# Extract each category from the total array
veggies = category_totals[0]
fruits = category_totals[1]
snacks = category_totals[2]
drinks = category_totals[3]

# Print the result
print(f"There are {veggies} vegetables, {fruits} fruits, {snacks} boxes of snacks, and {drinks} beverages in the two grocery bags.")
```

## Teamwork

```python
# Create an array for Ali's knocks (Mon to Fri)
ali = np.array([5, 4, 6, 4, 2])
print("Ali's knocks", ali)
```

```python
# Create an array for Ben's knocks (0 on Monday because he was out of town)
ben = np.array([0, 5, 5, 3, 4])
print("Ben's knocks", ben)
```

```python
# Create an array for Carla's knocks (0 on Wednesday because she was sick)
carla = np.array([3, 6, 0, 3, 4])
print("Carla's knocks", carla)
```

```python
# Compile all three individual arrays into a single array
all_knocks = np.array([ali, ben, carla])
print("all_knocks array")
print(all_knocks)
```

```python
# Calculate total for each person and the collective grand total
person_totals = np.sum(all_knocks, axis=1)
group_total = np.sum(all_knocks)

print("Totals by person (Ali, Ben, Carla):", person_totals)
print("Collective group total:", group_total)
```

```python
# 1. Get the total knocks for the group day by day 
daily_group_knocks = np.sum(all_knocks, axis=0)

# 2. Get the rolling total (cumulative sum) day by day
rolling_knocks = np.cumsum(daily_group_knocks)

# 3. Multiply by 0.5 to get estimated votes (2 knocks = 1 vote)
rolling_votes = rolling_knocks * 0.5

print("Rolling total of estimated votes day by day (Mon-Fri)")
print(rolling_votes)
```
