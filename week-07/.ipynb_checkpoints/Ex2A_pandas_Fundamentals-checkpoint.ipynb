# Exercise 2.A Fundamentals of pandas

```python
import numpy as np
import pandas as pd
```

## Series

```python
# Creating plain Python objects
aa = [1, 2, 3, 4]
ab = (5, 6, 7, 8)
ac = {9, 10, 11, 12}
ad = {'first': 13, 'second': 14, 'third': 15, 'fourth': 16}
```

```python
# Converting to Pandas Series 
ser1 = pd.Series(aa)
ser2 = pd.Series(ab)
ser3 = pd.Series(list(ac)) 
ser4 = pd.Series(ad)

# Printing each series 
print("ser1")
print(ser1)
print("\nser2")
print(ser2)
print("\nser3")
print(ser3)
print("\nser4")
print(ser4)
```

```python
# Try to create ser5 using plain literal numbers 
ser5 = pd.Series(17, 18, 19, 20)

# TypeError - Series() takes from 0 to 1 positional arguments but 4 were given
```

## DataFrames from Series

```python
# Creating a DataFrame using ser1
df1 = pd.DataFrame(ser1)
df1
```

```python
# Inserting ser2 at column position 1 and ser3 at column position 2
df1.insert(1, 'col2', ser2)
df1.insert(2, 'col3', ser3)
df1
```

```python
# Adding ser4 as a new column
df1['col4'] = ser4
df1

# Comment - ser4 was created from a dictionary with custom string keys,(first, second) 
# while df1 uses a standard numeric index (0, 1, 2, 3). Because the indexes do not match up, 
# Pandas could not align the data, resulting in a column filled entirely with NaN.
```

```python
# Adding row index 4 with the specific values 
df1.loc[4] = [5, 9, 12, 0]
df1
```

## Reading Data into a DataFrame

```python
# Displaying the first 5 rows
df2.head()
```

```python
# Displaying the last 10 rows
df2.tail(10)
```

```python
# Showing structural info of df2
df2.info()
```

### Facts Determined from df2 Info
* I can see the exact total number of rows (records) and columns in the dataset.
* It tells us the data types of each column (such as integers, floats, or text objects).
* It shows me if there are any missing or null values in my data columns, which helps me know if I need to clean anything.

```python
# Showing numerical summaries for df2
df2.describe()
```

```python
### Observations on df2 Describe Results
### What is NOT useful - The .describe() method treats everything like numbers, but my columns here contain movie titles, descriptions, 
### and categories. Because these are text data columns and not actual math numbers, the numerical summary statistics do not give us helpful business 
### insights for this specific dataset.
### What I can take away - It shows me that .describe() is best saved for datasets that have real numerical values like sales figures, prices,
### or quantities rather than text descriptions.
```

```python
# Reading the Northwind CSV file into df3
df3 = pd.read_csv('northwind_example.csv')
df3
```

```python
# Showing numerical summaries for the Northwind data
df3.describe()
```

```python
### Useful Insights from df3 Describe Results
### What is useful here - Unlike the previous movie data, df3 has real numeric math data like UnitPrice and Quantity. The .describe() method is 
### incredibly useful here because it gives me meaningful business insights, like the average price of the products, the maximum quantity ordered, 
### and the spread of the pricing.
### How it is different and why - This summary is much more useful than the df2 movie example because df3 contains actual numerical values meant
### for math calculations. The movie dataset was almost entirely descriptive text strings, which is why mathematical statistics did not make sense for it.
```

## Coding a DataFrame

```python
# Creating the test data dictionary 
test_data = {
    'ID': [1, 2, 3, 4, 5, 6, 7, 8],
    'Name': ['Alicia', 'Bob', 'Charlie', 'Dominic', 'Eve', 'Frank', 'Grace', 'Heidi'],
    'Age': [25, np.nan, 22, 19, 31, 35, np.nan, 18],
    'City': ['NY', 'LA', 'Chicago', 'Houston', 'Phoenix', 'Boston', 'Austin', 'San Diego'],
    'Score': [85, 92, -1, 65, 78, np.nan, 55, 90]
}
```

```python
# Create a DataFrame named df4 using test_data
df4 = pd.DataFrame(test_data)
df4
```

```python
# Exporting df4 to a CSV file named test_data.csv
df4.to_csv('test_data.csv', index=False)
```
