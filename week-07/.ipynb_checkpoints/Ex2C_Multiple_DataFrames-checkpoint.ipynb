# Exercise 2.C Working with multiple DataFrames

```python
import numpy as np
import pandas as pd
```

## Basic Aggregations

```python
# Create a DataFrame named df_sales
data = {
    'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North'],
    'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],
    'Price': [100, 200, 150, 100, 200, 150, 100, 200, 150]
}

df_sales = pd.DataFrame(data)
df_sales
```

```python
# Adding a new column Revenue by multiplying Units_Sold and Price
df_sales['Revenue'] = df_sales['Units_Sold'] * df_sales['Price']
df_sales
```

```python
# Calculating the total Units_Sold and total Revenue across all products and regions
total_units = df_sales['Units_Sold'].sum()
total_revenue = df_sales['Revenue'].sum()

print(f"Total Units Sold: {total_units}")
print(f"Total Revenue: {total_revenue}")
```

```python
# Calculating the average Units_Sold and average Revenue
avg_units = df_sales['Units_Sold'].mean()
avg_revenue = df_sales['Revenue'].mean()

print(f"Average Units Sold: {avg_units}")
print(f"Average Revenue: {avg_revenue}")
```

## Grouping Data

```python
# Grouping by Product and calculating the total Units_Sold and Revenue
product_totals = df_sales.groupby('Product')[['Units_Sold', 'Revenue']].sum()
product_totals
```

```python
# Grouping by Region and calculating the average Units_Sold and Revenue
region_averages = df_sales.groupby('Region')[['Units_Sold', 'Revenue']].mean()
region_averages
```

```python
# Grouping by both Product and Region to calculate the sum
combined_totals = df_sales.groupby(['Product', 'Region'])[['Units_Sold', 'Revenue']].sum()
combined_totals
```

## Sorting and Ranking Data

```python
# Sort df_sales by Units_Sold in ascending order
df_sales.sort_values(by='Units_Sold', ascending=True)
```

```python
# Sort df_sales by Revenue in descending order
df_sales.sort_values(by='Revenue', ascending=False)
```

```python
# Add a new column Revenue_Rank ranking the Revenue values in descending order
df_sales['Revenue_Rank'] = df_sales['Revenue'].rank(ascending=False)
```

```python
# Call df_sales to review the changes
df_sales
```

## Merging DataFrames

```python
# Create a new DataFrame named df_products
product_data = {
    'Product': ['A', 'B', 'C'],
    'Category': ['Electronics', 'Furniture', 'Appliances']
}
df_products = pd.DataFrame(product_data)
df_products
```

```python
# Merge df_sales with df_products on the Product column
df_merged = pd.merge(df_sales, df_products, on='Product')
```

```python
# Call df_merged to review the merged DataFrame
df_merged
```

```python
# Group by Category to calculate the total Revenue
category_revenue = df_merged.groupby('Category')['Revenue'].sum()
category_revenue
```
