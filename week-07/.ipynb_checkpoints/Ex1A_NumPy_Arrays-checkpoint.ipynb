```python
import numpy as np
```

## Looping Through Arrays

 ## Looping Through Arrays

```python
arrA = np.array([[11, 12, 13], [14, 15, 16]])
print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

The first loop iterates through the rows of the 2D array so it prints entire 1D arrays (rows) multiplied by 2. The second loop goes deeper 
into the second dimension iterating through each individual number inside those rows and printing them one by one.

```python
for x in arrA:
    for y in x:
        loop_result = x * 2
print(loop_result)

# Explanation
# This prints [28, 30, 32]. Because loop_result was assigned x * 2 it multiplied the entire current row (x) by 2 instead of the individual 
# number (y). Since it printed outside the loop I only see the result for the very last row processed.
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2))
print(stacked_array_2)

# Error Message - ValueError - all input arrays must have the same shape
# Explanation  
# I cannot stack these arrays because they have different shapes. 
# joined_array_1 is 4x3 but joined_array_2 is 2x6. To stack arrays their dimensions must match exactly.
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA), axis=0)
print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1, arrA), axis=1)
print(another_2)

# Error message - ValueError 
# This fails because the row counts don't match. joined_array_1 has 4 rows but arrA only has 2 rows. I cannot glue them side-by-side (axis=1) 
# because the heights are different.
```

```python
# Horizontal Stack (glue side-by-side)
h_result = np.hstack((arrA, arrA))
print("hstack result:\n", h_result)
# Comment - hstack combines arrays horizontally (columns change, rows stay same).

# Vertical Stack (glue top-to-bottom)
v_result = np.vstack((arrA, arrA))
print("\nvstack result:\n", v_result)
# Comment - vstack combines arrays vertically (adds more rows).

# Depth Stack (stack like a deck of cards into 3D)
d_result = np.dstack((arrA, arrA))
print("\ndstack result:\n", d_result)
# Comment - dstack combines arrays along a third axis (depth) making a 3D array.
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)
print(splitsky)

# Comment
# The result is a list containing 2 separate arrays. 
# Each of these new arrays has 2 dimensions with a shape of 2 rows and 3 columns.
```

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)

# Comment 
# The result is a list containing 3 separate arrays. 
# Each array is still 2-dimensional. The first array has 2 rows while the remaining two arrays have 1 row each.
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([[7, 6], [13, 5], [7, 3]])
arrB_sorted = np.sort(arrB)
print("Sorted Array:\n", arrB_sorted)
```

```python
print("Original arrB:\n", arrB)

z = arrB < 7
arrC = arrB[z]

print("\nBoolean Mask z:\n", z)
print("\nFiltered arrC (values < 7):\n", arrC)
```

```python
print("Original arrB:\n", arrB)

y = arrB >= 7
arrD = arrB[y]

print("\nBoolean Mask y:\n", y)
print("\nFiltered arrD (values >= 7):\n", arrD)
```
