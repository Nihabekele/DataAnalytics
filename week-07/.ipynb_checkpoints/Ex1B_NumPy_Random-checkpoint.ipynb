# Exercise 1.B NumPy Random

```python
import numpy as np
```

## Generating Random Numbers

```python
r_1 = np.random.randint(1, 21, size=2)
print("r_1:\n", r_1)
```

```python
r_2 = np.random.randint(1, 21, size=(1, 2))
print("r_2:\n", r_2)
```

```python
r_3 = np.random.random(size=(2, 1))
print("r_3:\n", r_3)
```

```python
r_4 = np.random.randint(20, 101, size=(4, 5))
print("r_4:\n", r_4)
```

```python
r_5 = np.random.random(size=(4, 5))
print("r_5:\n", r_5)
```

## Generating Random Choices from an Array

```python
students = np.array(['Alen', 'Biniam', 'Shafiqa', 'Daniel', 'Eleni', 'Filmon', 
                     'Grace', 'Hassan', 'Maya', 'Julia', 'Kevin', 'Lydia'])
print("Students Array:\n", students)
```

```python
call_on = np.random.choice(students)
print("Randomly Called On:", call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3, 4), replace=False)
print("Breakout Groups:\n", breakout_groups)

# Comment
# Yes, this is an effective way to assign breakout groups. 
# Setting replace=False ensures that every student gets placed into exactly one group of 4 without anyone being repeated or left out.
```

```python
# p list must have exactly 12 values matching 12 students
probabilities = [0.0, 0.3, 0.07, 0.07, 0.07, 0.07, 0.07, 0.07, 0.07, 0.07, 0.07, 0.07]

call_on_weighted = np.random.choice(students, size=5, p=probabilities)
print("Weighted Selections:", call_on_weighted)

# Explanation 
# I assigned a probability array p where Alen has a 0.0 (0%) chance of being picked,
# Biniam has a 0.3 (30%) chance and the remaining 10 students split the remaining 70% evenly (0.07 or 7% each). The entire list totals 1.0.
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))
print(bell_curve)
```
