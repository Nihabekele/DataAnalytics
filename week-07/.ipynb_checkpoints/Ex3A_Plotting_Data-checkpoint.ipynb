# Exercise 3.A Getting familiar with Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
# Creating the data variables
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
```

```python
# Plotting a basic line graph
plt.plot(x, y)
plt.show()
```

```python
# Plotting the line graph with add markers to the data points
plt.plot(x, y, marker='o')
plt.show()
```

```python
# Customizing the plot with labels and a title
plt.plot(x, y, marker='o')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

## Styling Plots

```python
# Plotting a dashed red line
plt.plot(x, y, color='red', linestyle='--')
plt.title('Dashed Red Line')
plt.show()
```

```python
# Plotting with a dashed red line and size 10 green square markers
plt.plot(x, y, color='red', linestyle='--', marker='s', markersize=10, markerfacecolor='green')
plt.title('Green Square Markers')
plt.show()
```

```python
# Experiment 1 - Blue dotted line with triangles
plt.plot(x, y, color='blue', linestyle=':', marker='^', markersize=12)
plt.title('Experiment 1 Blue Triangles')
plt.show()
```

```python
# Experiment 2 - Purple solid line with large stars
plt.plot(x, y, color='purple', linestyle='-', marker='*', markersize=15)
plt.title('Experiment 2 Purple Stars')
plt.show()
```

# Exercise 3.B More plotting practice

## Advanced Plotting Techniques

```python
# Creating a basic bar chart
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

plt.bar(products, sales)
plt.show()
```

```python
# Customizing the bar chart with skyblue color and specific width
plt.bar(products, sales, color='skyblue', width=0.5)

plt.title('Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```

```python
# Creating a basic scatter plot with random data
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y)
plt.show()
```

```python
# Customizing the scatter plot with varying sizes and colors
colors = np.random.rand(50)
sizes = np.random.rand(50) * 100

# Plotting with s (size), c (color array), and cmap (colormap)
plt.scatter(x, y, s=sizes, c=colors, cmap='viridis')
plt.title('Customized Scatter Plot')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.show()
```

## Subplots

```python
# Set up a figure that is wide enough to hold two charts side by side
plt.figure(figsize=(12, 5))

# First subplot sine wave 
# plt.subplot(Rows, Columns, Position)  1 row, 2 columns, 1st chart
plt.subplot(1, 2, 1)

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, color='blue')
plt.title('Sine Wave Chart')
plt.xlabel('X Values')
plt.ylabel('Y Values (sin(x))')

# Second subplot histogram 
# 1 row, 2 columns, 2nd chart
plt.subplot(1, 2, 2)

random_data = np.random.randn(1000)

plt.hist(random_data, color='skyblue', edgecolor='black')
plt.title('Histogram of 1000 Random Numbers')
plt.xlabel('Random Values')
plt.ylabel('Frequency (How many times)')

# plt.tight_layout() prevents the titles and labels from overlapping
plt.tight_layout()
plt.show()
```
