# Exercise 2.B Cleaning up data

```python
import numpy as np
import pandas as pd
```

```python
# Reading the saved CSV file back into a new DataFrame named df
df = pd.read_csv('test_data.csv')
df
```

```python
# Showing numerical statistics for df
df.describe()
```

## Filtering a DataFrame

```python
# Filtering the DataFrame to show only rows where Age is greater than 18
df[df['Age'] > 18]
```

```python
# Filtering the DataFrame for Scores greater than 50 AND less than 100
df[(df['Score'] > 50) & (df['Score'] < 100)]
```

## Modifying a DataFrame

```python
# Create df_cleaned and drop rows where Score is less than 0
df_cleaned = df.drop(df[df['Score'] < 0].index)
df_cleaned
```

```python
# Remove the City column using drop
df_cleaned = df_cleaned.drop(columns=['City'])
df_cleaned
```

```python
# Replace missing Age values with the mean Age
mean_age = df_cleaned['Age'].mean()
df_cleaned['Age'] = df_cleaned['Age'].fillna(mean_age)
df_cleaned
```

```python
# Replace Score value of 0 with the mean Score
mean_score = df_cleaned['Score'].mean()
df_cleaned['Score'] = df_cleaned['Score'].replace(0, mean_score)
df_cleaned
```

```python
# Remove any remaining rows that contain NaN values
df_cleaned = df_cleaned.dropna()
df_cleaned
```

```python
# Apply values from ID as index labels and remove the ID column
df_cleaned = df_cleaned.set_index('ID')
df_cleaned
```
