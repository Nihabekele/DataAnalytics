# Exercise 2.B working with datetime functions.

## Lab 2

```python
import datetime
```

# Today

```python
today = datetime.datetime.now()
```

```python
formatted_date = today.strftime("%A, %B %d, %Y")
print(f"Today is {formatted_date}")
```

# My birthday

```python
# 9. Define your birthday (Year, Month, Day)
my_birthday = datetime.date(1998, 5, 20)
```

```python
# 10. Format as MM/DD/YYYY
formatted_bday = my_birthday.strftime("%m/%d/%Y")
print(f"My birthday is {formatted_bday}")
```

# Next 90 Days

```python
# 12. Calculate 90 days from today
ninety_d = today + datetime.timedelta(days=90)
```

```python
# 13. Format as Month, Day,Year
future_date = ninety_d.strftime("%B %d, %Y")
print(f"90 days from today is {future_date}")
```

# Time for Dinner

```python
# 15. Define dinner time (6:30 PM is 18:30)
dinner_time = datetime.time(18, 30)
```

```python
# 16. Format as HH:MM PM
formatted_dinner = dinner_time.strftime("%I:%M %p")
print(f"Let’s meet for dinner at {formatted_dinner}")
```
