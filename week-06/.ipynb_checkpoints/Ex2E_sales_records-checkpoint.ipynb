### Exercise 2.E Using functions for analyzing data
This notebook analyzes a small sales dataset using Python functions.
Each function takes the full dataset as its first argument.

```python
sales_data = [
    ('Laptop', 'East', 1200.00),
    ('Phone', 'West', 850.00),
    ('Tablet', 'East', 400.00),
    ('Monitor', 'South', 650.00),
    ('Keyboard', 'West', 75.00),
    ('Laptop', 'South', 1150.00),
    ('Phone', 'East', 900.00),
    ('Tablet', 'West', 425.00),
    ('Monitor', 'East', 680.00),
    ('Keyboard', 'South', 80.00),
    ('Laptop', 'West', 1300.00),
    ('Phone', 'West', 820.00),
]
```

## Lab 1. Total sales by region

```python
def region_total(data, region):
    return sum(amt for _, r, amt in data if r == region)
```

```python
print(region_total(sales_data, 'East'))
print(region_total(sales_data, 'West'))
print(region_total(sales_data, 'South'))
```

## Lab 2. Count of sales by region

```python
def region_count(data, region):
    # This sums the number 1 for every row where the region matches
    return sum(1 for _, r, _ in data if r == region)
```

```python
print(region_count(sales_data, 'East'))
print(region_count(sales_data, 'West'))
print(region_count(sales_data, 'South'))
```

## Lab 3. Average sale amount by region

```python
def region_average(data, region):
    # Get the count first to check if it's zero
    count = region_count(data, region)
    
    # If count is 0, return 0 to avoid a "Division by Zero" error
    if count == 0:
        return 0
        
    # get the total and divide by the count
    total = region_total(data, region)
    return total / count
```

```python
print(f"East Average:  {region_average(sales_data, 'East')}")
print(f"West Average:  {region_average(sales_data, 'West')}")
print(f"South Average: {region_average(sales_data, 'South')}")
print(f"North Average: {region_average(sales_data, 'North')}") 
```

## Lab 4. Filter by sale amount

```python
def sales_above(data, threshold):
    return [record for record in data if record[2] > threshold]
```

```python
high_value_sales = sales_above(sales_data, 500)

# Print each record individually
print("Sales above $500:")
for record in high_value_sales:
    print(record)
```

# Bonus

```python
def sales_below(data, threshold):
    return [record for record in data if record[2] < threshold]
```

```python
# Call the function for sales less than 500
low_value_sales = sales_below(sales_data, 500)

# Print each record individually
print("Sales below $500:")
for record in low_value_sales:
    print(record)
```

## Lab 5. Top and bottom sale

```python
def top_sale(data):
    # Returns the tuple with the highest value at index 2
    return max(data, key=lambda s: s[2])

def bottom_sale(data):
    # Returns the tuple with the lowest value at index 2
    return min(data, key=lambda s: s[2])

# Testing the functions
print(f"Top Sale: {top_sale(sales_data)}")
print(f"Bottom Sale: {bottom_sale(sales_data)}")
```

## The top sale was a Laptop in the West region for $1300.0.    

## The bottom sale was a Keyboard in the West region for $75.0.

## Lab 6. Apply a discount

```python
def apply_discount(data, discount_pct):
    return list(map(lambda s: (s[0], s[1], s[2] * (1 - discount_pct)), data))
```

```python
# 1. Apply a 15% discount
discounted_sales = apply_discount(sales_data, 0.15)

# 2. Print first three records BEFORE
print("First 3 records (Original):")
for record in sales_data[:3]:
    print(record)

print("-" * 30)

# 3. Print first three records AFTER
print("First 3 records (15% Discount):")
for record in discounted_sales[:3]:
    print(record)
```

## Lab 7. Unique regions

```python
def unique_regions(data):
    return sorted(set(r for _, r, _ in data))

# Testing the function
regions = unique_regions(sales_data)
print(f"Unique Regions: {regions}")
```

## This function is useful because it automatically finds all the regions in the data without me having to type them manually. It allows the program to stay up to date even if new regions are added to the list later.

# Lab 8. Region summary

```python
def region_summary(data, region):
    total = region_total(data, region)
    count = region_count(data, region)
    avg = region_average(data, region)
    
    # Filter data for just this region to find the top sale
    region_data = [record for record in data if record[1] == region]
    
    # Print the report
    print(f"=== {region} Region ===")
    print(f"Records: {count}")
    print(f"Total Sales: ${total:,.2f}")
    print(f"Average Sale: ${avg:,.2f}")
    
    if count > 0:
        top = top_sale(region_data)
        # top[0] is product, top[2] is amount
        print(f"Top Sale: {top[0]} — ${top[2]:,.2f}")
    else:
        print("Top Sale: N/A")
    print() 
    
region_summary(sales_data, "East")
region_summary(sales_data, "West")
```

## Lab 9. Full summary

```python
all_regions = unique_regions(sales_data)

# Loop through each region
for reg in all_regions:
    region_summary(sales_data, reg)
```
