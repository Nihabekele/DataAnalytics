##  Lab 3 Exercise 2C Tax Calculation Functions

```python
# This lab demonstrates the use of Python functions to automate financial calculations. 
# Arithmetic operations for fixed tax rates (Social Security and Medicare).
# Conditional logic (if/elif/else) to determine variable tax rates based on withholding codes.
# Parameter handling to process different employee data dynamically.
```

```python
#2a.
def get_soc_sec_tax(gross_pay):
    # Tax rate is 6.2% (0.062)
    return gross_pay * 0.062
```

```python
#b
def get_medicare_tax(gross_pay):
    # Tax rate is 1.45% (0.0145)
    return gross_pay * 0.0145
```

```python
#c
def get_federal_tax(gross_pay, withholding_code):
    if withholding_code == 0:
        rate = 0.23
    elif withholding_code == 1:
        rate = 0.21
    elif withholding_code == 2:
        rate = 0.195
    elif withholding_code == 3:
        rate = 0.185
    else:
        # For code 4+, start at 18% and subtract 0.5% for every code above 4
        extra_withholding = withholding_code - 4
        rate = 0.18 - (extra_withholding * 0.005)
        
    return gross_pay * rate
```

```python
#3
pay = 750
code = 0
print(f"Person 1 - Social Security: ${get_soc_sec_tax(pay):.2f}")
print(f"Person 1 - Medicare: ${get_medicare_tax(pay):.2f}")
print(f"Person 1 - Federal Tax: ${get_federal_tax(pay, code):.2f}")
```
