## working with defining custom functions.

```python
# Create the function with three parameters
def favorite_things(name, movie, music):
    line1 = f"My name is {name}."
    line2 = f"My favorite movie is {movie}."
    line3 = f"My favorite musician is {music}."
    return f"{line1}\n{line2}\n{line3}"

# Call the function and print the result
print(favorite_things("Hana", "Straw", "Rihana"))
```

```python
# Function that returns the purpose sentence
def why_im_here(purpose):
    return purpose

my_reason = "I joined Year Up United to gain technical skills and start a career in Data Analytics."
print(why_im_here(my_reason))
```

```python
# Function with two parameters - place and why
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."

print(favorite_place("the beach", "the sound of the waves is very relaxing"))

print(favorite_place("Ethiopia", "it has beautiful mountains and amazing food"))
```
