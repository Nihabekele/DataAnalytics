# This notebook covers NumPy basics converting lists to 1D, 2D, and 3D arrays, inspecting data types (dtype), and manipulating shapes using the transpose attribute, indexing, and slicing.

```python
import numpy as np
```

## Array Basics

```python
fruits = ["apple", "banana", "cherry", "date"]
```

```python
arr_1 = np.array(fruits)
print(arr_1)
```

```python
print(arr_1.dtype)

# The result is <U6. 
# The '<' means little-endian.
# The 'U' means Unicode.
# The '6' means the longest word in the array has 6 characters.
```

```python
# I use nested brackets for 2D arrays: [[row1], [row2]]
arr_2 = np.array([[10, 11, 12], [13, 14, 15]])
print(arr_2)
```

```python
# Print the transposed version
print(arr_2.T)
```

```python
print(arr_2)
# No, arr_2.T does not alter the original array. 
# It only shows a transposed view. The original variable remains the same.
```

```python
arr_3 = np.array([[1, 3, 5], [2, 4, 6]])
print(arr_3)
```

```python
# Create arr_4 using the specific elements
arr_4 = np.array([arr_3[0, 2], arr_3[1, 1]])
print(arr_4)
```

```python
# Shape (2, 2, 2) means 2 layers, 2 rows, 2 columns
arr_5 = np.ones((2, 2, 2))
print(arr_5)
```
