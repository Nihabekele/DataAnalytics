## Lab 3. Employee data
This lab exercise makes updates to information extracted from
an employee database.

```python
# 3.Creating a list of tuples containing ID, Dept, Name, and Salary
hr_list = [
    ('0123', 'HR', 'Rebecca Yang', '69000'),
    ('0121', 'IT', 'Mark Blick', '67000'),
    ('0068', 'IT', 'Kahna Larsen', '112000'),
    ('0234', 'OPS', 'Jim Smith', '54000')
]
```

```python
# 4a. Replacing the tuple at index 1 with updated information
hr_list[1] = ('0121', 'IT', 'Mark Blick-Hawley', '67000')
```

```python
# 4b. Printing the list to see Mark's new name
print(hr_list)
```

```python
# 5a. Replacing the tuple at index 3 with the new Dept (CS) and new Salary (60000)
hr_list[3] = ('0234', 'CS', 'Jim Smith', '60000')
```

```python
# 5b. Printing the final list to ensure all changes (Mark and Jim) are there
print(hr_list)
```

```python
# 6. Displaying updated contents with pipe separation and formatted salary
for employee in hr_list:
    # Unpack the tuple into variables
    emp_id, dept, name, salary = employee
    
    # CAST the salary from string to int, then format with a comma
    formatted_salary = "{:,}".format(int(salary))
    
    # Print using the pipe character | with spaces
    print(f"{emp_id} | {dept} | {name} | {formatted_salary}")
```
