## Lab 1. Working with names – parsing strings

```python
test_name = 'John Doe'
```

```python
def parse_and_display(name):
    for i in range(len(name)):
        if name[i] == ' ':
            # Use slicing: [:i] gets everything BEFORE the space
            # [i+1:] gets everything AFTER the space
            first = name[:i]
            last = name[i+1:]
            
            print(f"Full name: {name}")
            print(f"First name: {first}")
            print(f"Last name: {last}")

# call the function to see it work
parse_and_display("John Smith")
```

```python
test_list = ['Grace Flores', 'JB Oinonen', 'Ken Penn']
```

```python
parse_and_display(test_list[0])
```

```python
def parse_and_display2(name):
    # .split() turns "Megan Thee Stallion" into ['Megan', 'Thee', 'Stallion']
    parts = name.split()
    count = len(parts)
    
    print(f"Full name: {name}")
    
    if count == 1:
        print(f"First name: {parts[0]}")
        print("Middle name: N/A")
        print("Last name: N/A")
        
    elif count == 2:
        print(f"First name: {parts[0]}")
        print("Middle name: N/A")
        print(f"Last name: {parts[1]}")
        
    elif count == 3:
        print(f"First name: {parts[0]}")
        print(f"Middle name: {parts[1]}")
        print(f"Last name: {parts[2]}")
    
    print("-" * 20) # This adds a little line between results

# Testing the Challenge names:
parse_and_display2('Lorde')
parse_and_display2('Billie Eilish')
parse_and_display2('Megan Thee Stallion')
```
