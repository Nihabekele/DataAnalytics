# Exercise 2.A Using built-in functions

## Lab 2. Parsing strings – part codes

#supplier:productcode-size

```python
def get_supplier_code(part_string):
    # .split(':') creates a list: [everything before ':', everything after ':']
    # I return index [0] to get the supplier name.
    return part_string.split(':')[0]
```

```python
def get_product_num(part_string):
    # First, I split at the colon and take the second part [1] -> "123-L"
    # Then, I split that result at the dash and take the first part [0] -> "123"
    return part_string.split(':')[1].split('-')[0]
```

```python
def get_size(part_string):
    # Index [1] gives us the size code (like 'L', 'M', or '12').
    return part_string.split('-')[1]
```

```python
# Create variables to store the example part 
part_a = "ACME:123-L"
part_b = "DI:12345-M"
part_c = "ACE:1-12"
```

```python
# Call each function 
supplier = get_supplier_code(part_a)
product = get_product_num(part_a)
size = get_size(part_a)

print(f"For {part_a}, the supplier is {supplier}. Part number {product} is available in size {size}.")
```

```python
# Repeat the process for the second example (DI)
supplier = get_supplier_code(part_b)
product = get_product_num(part_b)
size = get_size(part_b)

print(f"For {part_b}, the supplier is {supplier}. Part number {product} is available in size {size}.")
```

```python
# Repeat the process for the third example (ACE)
supplier = get_supplier_code(part_c)
product = get_product_num(part_c)
size = get_size(part_c)

print(f"For {part_c}, the supplier is {supplier}. Part number {product} is available in size {size}.")
```

```python
# Bonus Create a list containing all three part codes
all_parts = ["ACME:123-L", "DI:12345-M", "ACE:1-12"]
```

```python
# Loop through each part in the list
for part in all_parts:
    supplier = get_supplier_code(part)
    product = get_product_num(part)
    size = get_size(part)
    
    print(f"For {part}, the supplier is {supplier}. Part number {product} is available in size {size}.")
```
